```python
import numpy as np
import pandas as pd 
```

```python
df = pd.DataFrame (
    [
        [8 , 8, 4],
        [7, 9 , 5],
        [6 , 10, 6],
        [5 , 12 , 7]
    ]
    ,
    columns = ['cgpa' , "profile_score","lpa"]
)
```

```python
import tensorflow
from tensorflow import keras 
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
model.add(
    Dense(
        2 , activation = 'linear' , input_dim = 2 
    )
)
model.add(
    Dense(
        1 , activation = 'linear'
    )
)
model.summary()
```

```python
model.get_weights()
```

```python
new_weights = [np.array([[0.1 ,  0.1 ], [ 0.1 , 0.1]], dtype=np.float32), #w1
                np.array([0., 0.], dtype=np.float32), #b1
                np.array([[0.1],[0.1]], dtype=np.float32), #w2 
                np.array([0.], dtype=np.float32)] #b2
model.set_weights(new_weights)
model.get_weights()
```

```python
optimizer = keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss = "mean_squared_error" , optimizer=optimizer)
```

```python
X = df.iloc[ : ,0 : -1].values
y = df['lpa'].values
model.fit(X , y , epochs= 75 , verbose= 1 , batch_size= 1) # verbose is just output lenght , batchsize is taking 1 row of data at a time
```
